# Introduction to Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

Make a plot using Matplotlib

```python
x = np.array([2020, 2021, 2022, 2023, 2024]) 
y = np.array([50, 55, 53, 60, 65])

# if only one param is provided it's the y values

plt.plot(x, y, marker='o', linestyle='-', color='b')
plt.title('Yearly Data Trend')
plt.xlabel('Year')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

cutomizing plots

```python
# set marker, linstyle, color, markersize, linewidth, markerfacecolor, markeredgewidth, markeredgecolor
y2 = np.array([60, 62, 61, 65, 70])
plt.plot(y, 
         marker='x', 
         linestyle='--',
            color='r',
         markersize=10,
         linewidth=2,
         markerfacecolor='yellow',
         markeredgewidth=2,
         markeredgecolor='green')

plt.title('Data Trend Over Indices')
plt.xlabel('Index')
plt.ylabel('Value')
```

Plots with multiple ys

```python


# Sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

# Plot multiple lines
plt.plot(x, y1, label='sin(x)', color='blue')
plt.plot(x, y2, label='cos(x)', color='red')
plt.plot(x, y3, label='sin(x)*cos(x)', color='green')

# Add labels, title, legend
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Line Plot Example')
# add legend to plot
plt.legend()

# Show plot
plt.show()
```

Creating a dictionary to store customizations

```python
plot_customization = dict(
    marker = 'o',
    linestyle = '-.',
    markersize = 8,
    linewidth = 2,
    markerfacecolor = 'orange',
    markeredgewidth = 1.5,
    markeredgecolor = 'black'
)
# ** means to unpack the dictionary into keyword arguments
# you can also just pass in different params
plt.plot(x, y, color='blue', **plot_customization)
plt.plot(x, y2, color='purple', **plot_customization)
plt.title('Customized Plot Example')
```

Full Customization of Plots, Labels, Grids and ticks

```python
x = np.array([1, 2, 3, 4, 5])
y1 = np.array([2, 3, 5, 7, 11])
y2 = np.array([1, 4, 6, 8, 10])
y3 = np.array([5, 3, 4, 2, 1])

# make curtomized labels
plt.title('Multiple Lines with Different Styles', 
          fontsize=14, 
          fontweight='bold',
          color='darkgreen', 
          loc='center',
          family='serif')
x_y_label_style = dict(
    fontsize=12,
    fontweight='bold',
    color='brown',
    family='sans-serif'
)
plt.plot(x, y1, marker='o', linestyle='-', color='blue', label='Line 1')
plt.plot(x, y2, marker='s', linestyle='--', color='red',
            label='Line 2') 
plt.plot(x, y3, marker='^', linestyle=':', color='green', label='Line 3')
plt.xlabel('X-axis', **x_y_label_style)
plt.ylabel('Y-axis', **x_y_label_style)
plt.legend()
# adds a grid to the plot and customizes its appearance
plt.grid(True, linestyle='--', color='lightgray', alpha=0.7)

#customize ticks
# plt.xticks(ticks=x, labels=['One', 'Two', 'Three', 'Four', 'Five'], rotation=45)
# plt.yticks(ticks=[1, 3, 5, 7, 9, 11], labels=['One', 'Three', 'Five', 'Seven', 'Nine', 'Eleven'])

# you can also select both axes = 'both' 
plt.tick_params(axis='x', direction='inout', length=6, width=2, colors='purple')
plt.tick_params(axis='y', direction='inout', length=6, width=2, colors='orange')
plt.show()
```

Customizing Barcharts

<small>used to compare categories of data </small>

```python
# x-axis
categories = np.array(['mango', 'banana', 'cherry', 'date', 'fig'])
# y-axis
values = np.array([5, 7, 3, 8, 4])
bar_style = dict(
    color='skyblue',
    edgecolor='black'
)
plt.title('Fruit Categories Bar Chart', fontsize=16, fontweight='bold', color='darkblue', loc='center', family='serif')
plt.bar(categories, values, **bar_style)
plt.xlabel('Fruit Types', fontsize=12, fontweight='bold', color='brown', family='sans-serif')
plt.ylabel('Quantity', fontsize=12, fontweight='bold', color='brown', family='sans-serif')
plt.grid(axis='y', linestyle='--', color='lightgray', alpha=0.7)
plt.show()

# you can use a barplot using plt.barplot to load from pandas dataframe directly
```

Customizing Pie Charts

```python
# like bargraph shows the proportions of different categories
categories = np.array(['Freshmen', 'Sophomore', 'Junior', 'Senior'])
values = np.array([120, 150, 100, 80])

plt.title('Student Enrollment by Year', fontsize=16, fontweight='bold', color='darkblue', loc='center', family='serif')
plt.pie(values, labels=categories,
        # sets the format of the percentage display on each slice
        autopct='%1.1f%%', 
        # rotates the start of the pie chart by the specified angle in degrees
        startangle=140, 
        colors=['gold', 'lightcoral', 'lightskyblue', 'lightgreen'], 
        #you can detach a bit to highight using explode)
        explode=(0.1, 0, 0, 0), 
        shadow=True
)

plt.show()
```

Customizing using scatterplots

<small> Shows the relationship between two variables , helps to idenitfy a correlation </small>

```python
# kinda like the plot
hours_of_study = np.array([1, 1, 3, 3, 5, 6, 7, 8, 9, 10])
scores = np.array([55, 60, 65, 70, 75, 80, 85, 90, 95, 100])

hours_of_study_2 = np.array([2, 4, 5, 6, 7, 8, 9])
scores_2 = np.array([58, 63, 68, 73, 78, 83, 88])

plt.scatter(hours_of_study, scores, 
            color='blue',
            marker='o',
            s=100,  # size of markers
            alpha=0.7,  # transparency
            edgecolor='black'  # border color of markers
              )
plt.scatter(hours_of_study_2, scores_2, 
            color='red',
            marker='^',
            s=100,  # size of markers
            alpha=0.7,  # transparency
            edgecolor='black'  # border color of markers
                )

plt.legend(['Group 1', 'Group 2'])
plt.title('Study Hours vs. Scores', fontsize=16, fontweight='bold', color='darkblue', loc='center', family='serif')
plt.xlabel('Hours of Study', fontsize=12, fontweight='bold', color='brown', family='sans-serif')
plt.ylabel('Scores', fontsize=12, fontweight='bold', color='brown', family='sans-serif')
plt.grid(True, linestyle='--', color='lightgray', alpha=0.7)
plt.show()
```

Historgrams

<small>a visual distribution of quatitative data (like normal distribution), gorup vlues into bins or intervals  and count how many falls in each range </small>

```python
#loc is the median, scale is the standard deviation, size is the number of samples
#sd is how much variation from the average or the center (loc)
scores = np.random.normal(loc=75, scale=10, size=100)
# clip to a certain range
scores = np.clip(scores, 0, 100)

# bins are the numbers of intervals
plt.hist(scores, bins=10, color='lightblue', edgecolor='black', alpha=0.7)
plt.title('Distribution of Scores', fontsize=16, fontweight='bold', color='darkblue', loc='center', family='serif')
plt.xlabel('Scores', fontsize=12, fontweight='bold', color='brown', family='sans-serif')
plt.ylabel('Frequency', fontsize=12, fontweight='bold', color='brown', family='sans-serif')
plt.grid(True, linestyle='--', color='lightgray', alpha=0.7)
plt.show()
```

Subplots in MatplotLib

FIgure vs Ax

<small>
Figure - the entire canvas
</small>
<br/>
<small>Ax - a single plot (subplot) </small>

```python
x = np.array([1,2,3,4,5])

# figsize = (rows, columns)
# subplot = plt.subplots(3,2)
# has two arrays: one for figure and one for axes
# 2 rows, 2 columns
fig, axes = plt.subplots(2,2)
# sekect axis using axes[row, column]
axes[0,0].set_title('Line Plot')
axes[0,0].plot(x, x**2, marker='o', color='blue')
axes[0,0].set_title('y = x^2')

axes[0,1].set_title('Bar Chart')
axes[0,1].bar(x, x*3, color='orange')
axes[0,1].set_title('y = 3x')

axes[1,0].set_title('Scatter Plot')
axes[1,0].scatter(x, x+2, color='green')
axes[1,0].set_title('y = x + 2')

axes[1,1].set_title('Histogram')
data = np.random.randint(1, 10, size=50)
axes[1,1].hist(data, bins=5, color='purple', edgecolor='black')
axes[1,1].set_title('Random Data Histogram')

plt.show()
```

Integrating MatplotLib with Pandas

```python
# for a bar chart
import pandas as pd

df = pd.read_csv('pkemon') 
# you an use a horizontal bar chart using plt.barh

type_count = df['Type 1'].value_counts()

plt.bar(type_count.index, type_count.values, color='lightblue', edgecolor='black')
plt.title('Pokemon Type Distribution', fontsize=16, fontweight='bold', color='darkblue', loc='center', family='serif')
plt.xlabel('Pokemon Types', fontsize=12, fontweight='bold', color='brown', family='sans-serif')
plt.ylabel('Count', fontsize=12, fontweight='bold', color='brown', family='sans-serif')
plt.xticks(rotation=45)
plt.show()
```
