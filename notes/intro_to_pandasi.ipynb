Casting Types

🐼 Pandas Data Type Casting Cheat Sheet

🔄 Common `astype()` Targets

| Data Type         | Pandas dtype string         | Notes |
|-------------------|-----------------------------|-------|
| Integer (default) | `int`, `int64`, `int32`, `int16`, `int8` | Use fixed-width for memory control |
| Unsigned Integer  | `uint64`, `uint32`, `uint16`, `uint8`    | No negative values allowed |
| Float             | `float`, `float64`, `float32`, `float16` | Precision varies by width |
| Boolean           | `bool`                      | Converts non-zero → `True`, zero/NaN → `False` |
| String            | `str`, `"string"` (new pandas string dtype) | `"string"` is preferred over `object` for text |
| Object            | `object`                    | Generic fallback, often for mixed types |
| Category          | `"category"`                | Efficient for repeated labels |
| Datetime          | `"datetime64[ns]"`          | Nanosecond precision timestamps |
| Timedelta         | `"timedelta64[ns]"`         | Differences between datetimes |
| Complex           | `complex`, `complex128`, `complex64` | For complex numbers |

Filtering Data based on conditions

````python
bios = pd.read_csv('./warmup_data/bios.csv')

# filtering usign loc 
# rows, columns
bios.loc[bios['height_cm'] > 180, ['name', 'height_cm']]

# shorthand - 2d array
bios[bios['height_cm'] > 180][['name', 'height_cm']]

#multiple conditions - parentheses needed
# use logical and & or |
print(bios[(bios['height_cm'] > 180) & (bios['born_country'] == 'USA')][['name', 'height_cm', 'born_country']])
````

Filtering Data based on specific thigns

````python
# string methods
# you can also use regular expressions
bios.loc[bios['name'].str.lower().str.contains('john|patrick')][['name']]
````

Powerful filtering using regex in pandas

````python
#chose to drop na matches using na=False

#find athletes born in cities that start with a vowel
vowel_cities = bios[bios['born_city'].str.lower().str.contains(r'^[aeiouAEIOU]', na=False, case=False)][['name', 'born_city']]
print("---VOWEL CITIES---\n")
print(vowel_cities.head(2))


#Find athletes whose names contain exctly 2 vowels 

two_vowels = bios[bios['name'].str.contains(r'^(?:[^aeiouAEIOU]*[aeiouAEIOU]){2}[^aeiouAEIOU]*$', na=False, case=False)][['name']]
print("---TWO VOWELS---\n")
print(two_vowels.head(2))

#Find athletes with repeated consecutive letters in their names
repeated_letters = bios[bios['name'].str.contains(r'(.)\1', na=False)][['name']]
print("---REPEATED LETTERS---\n")
print(repeated_letters.head(2))

#Find athletes with names ending in 'son' or 'sen'
ending_son_sen = bios[bios['name'].str.contains(r'son$|sen$', na=False=)][['name']]
print("---ENDING SON OR SEN---\n")
print(ending_son_sen.head(2))

#Athletes born in year starting with 19
nineties_born = bios[bios['born_date'].str.contains(r'^19', na=False)][['name', 'born_date']]
print("---NINETIES BORN---\n")
print(nineties_born.head(2))

# you can turn off regex by setting regex=False
````

Defining a function to handle data

````python
def categorize_athlete(row): 
    if row['height_cm'] < 180:
        return 'Lightweight'
    elif 180 <= row['height_cm'] <= 200:
        return 'Middleweight'
    else:
        return 'Heavyweight'
    
#axis is the direction, 0 for columns, 1 for rows
bios['weight_class'] = bios.apply(categorize_athlete, axis=1)
bios[['name', 'height_cm', 'weight_class', 'height_cat']].head()
````

What to do with Null Values - Real World

FillNA with Mode, Median, Mean

````python
# fill with median value of the column
#column: value
coffee.fillna({'Units Sold': coffee['Units Sold'].median()})

# you can use mean, median, mode or a specific value like based on a coffee type
````

Fill NA by letting the programs recognize patterns (interpolate)

````python
# if a better fit is needed use interpolate based on items BESIDES the VALUES
coffee.fillna(coffee['Units Sold'].interpolate())
````

DropNAS

````python
coffee.dropna(subset=['Units Sold'])
#drops the full entire row , make sure youre basing off a column that matters
````

Get Only Rows with/without NAs

````python
# get rows with missing 'Units Sold'
coffee[coffee['Units Sold'].isna()]

# get all rows without missing 'Units Sold'
coffee[coffee['Units Sold'].notna()]
````

Using GroupBy

````python
coffee = pd.read_csv('./warmup_data/coffee.csv')
coffee.head()
````

Pivots

````python
pivot = coffee.pivot_table(index='Coffee Type', columns='Day', values='Units Sold', aggfunc='sum', fill_value=0)
# reshaping data for better analysis

# index → rows (here: "Coffee Type")

# columns → new column headers (here: "Day")

# values → the data you want to summarize (here: "Units Sold")

# aggfunc → how to aggregate values if there are duplicates (here: sum)

# fill_value → replaces missing values with a default (here: 0)

# you can treat pivots like normal dataframes and run 'sum' or other methods on them
pivot.sum(axis=1)  # Sum across rows (total units sold per Coffee Type)
#reset_index to turn back to normal dataframe
````
