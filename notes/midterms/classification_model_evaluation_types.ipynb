
### 1. Accuracy
Definition: Ratio of correctly predicted instances to total instances.

Usage: Best for balanced datasets.

Function: sklearn.metrics.accuracy_score(y_true, y_pred)



### 2. Confusion Matrix
Definition: Tabular summary of predictions vs. actuals.

Usage: Foundation for other metrics; shows TP, FP, TN, FN.

Function: sklearn.metrics.confusion_matrix(y_true, y_pred)




### 3. Precision, Recall, F1-Score
Precision: Correct positive predictions / total predicted positives.

Recall (Sensitivity): Correct positive predictions / total actual positives.

F1-Score: Harmonic mean of precision and recall.

Usage: Critical for imbalanced datasets (e.g., fraud detection).

Function: sklearn.metrics.precision_score, recall_score, f1_score

Note that Precision and Recall are usually tradeoffs

Precision is how much correct did i get from my classification, recall is how much did i classify from the actual data



The f1 score ensure that both recall and precions are kept high for good models


### 4. Classification Report
Definition: Combined summary of precision, recall, F1-score, and support.

Usage: Quick overview of performance across classes.

Function: sklearn.metrics.classification_report(y_true, y_pred)

### 5. ROC Curve & AUC
ROC Curve: Plots True Positive Rate vs. False Positive Rate.

AUC (Area Under Curve): Measures overall ability to distinguish classes.

Usage: Best for probabilistic classifiers (e.g., Logistic Regression).

Function: roc_curve, roc_auc_score


<small>The ROC curve (Receiver Operating Characteristic curve) is a tool used to evaluate the performance of a binary classification model. Its purpose is to show how well a model can distinguish between two classes (e.g., positive vs. negative, spam vs. not spam) across different decision thresholds.

🔍 Key Points
Axes:

X-axis: False Positive Rate (FPR) = proportion of negatives incorrectly classified as positives.

Y-axis: True Positive Rate (TPR) = proportion of positives correctly classified.

Purpose:

Visualizes the trade-off between sensitivity (recall) and specificity.

Helps compare different models or thresholds.

Provides a threshold-independent measure of performance.

AUC (Area Under the Curve):

A single number summarizing the ROC curve.

Ranges from 0 to 1:

0.5 → no better than random guessing.

1.0 → perfect classification.

Higher AUC = better model discrimination.

⚖️ Why It’s Useful
It avoids bias toward a particular threshold (unlike accuracy).

It’s especially valuable when dealing with imbalanced datasets, where accuracy can be misleading.

It helps project managers, data scientists, and engineers decide whether a model is robust enough for deployment.

Think of it as a diagnostic lens: instead of asking “How often is the model right?”, it asks “How well does the model separate the two classes across all possible cutoffs?”
