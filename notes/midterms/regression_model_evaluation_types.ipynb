| Metric                   | Definition                                      | Strengths                                | Weaknesses                              | Best Use Case                           | Interpretation (Scale)                  |
|---------------------------|------------------------------------------------|------------------------------------------|------------------------------------------|------------------------------------------|------------------------------------------|
| **MAE**                   | Avg. absolute error                            | Easy to interpret, robust                | Treats all errors equally                 | General regression tasks                 | Lower is better (0 = perfect)           |
| **MSE**                   | Avg. squared error                             | Penalizes large errors                   | Sensitive to outliers                     | When large errors are critical           | Lower is better (0 = perfect)           |
| **RMSE**                  | √MSE                                           | Same units as target                     | Still sensitive to outliers               | Model comparison, intuitive scale        | Lower is better (0 = perfect)           |
| **MSLE**                  | Avg. squared log error                         | Handles exponential growth               | Requires positive target values           | Forecasting with wide value ranges       | Lower is better (0 = perfect)           |
| **R² Score**              | Variance explained                             | Standard goodness-of-fit measure         | Can be negative, misleading for small data| Overall model performance                | 1 = perfect fit, 0 = poor fit, <0 = worse than baseline |
| **Explained Variance**    | Variance explained by predictions              | Similar to R², less strict               | Less informative than R²                  | Quick variance check                     | 1 = perfect, 0 = worst                  |
| **Median Absolute Error** | Median of absolute errors                      | Robust to outliers                       | Less sensitive to overall error magnitude | Datasets with heavy-tailed distributions | Lower is better (0 = perfect)           |


### 1. Mean Absolute Error (MAE)
Definition: Average of absolute differences between predicted and actual values.

Function: sklearn.metrics.mean_absolute_error(y_true, y_pred)

Usage: Easy to interpret, less sensitive to outliers. 
"how wrong was the prediction on average" 
--- Reason why it isn't used is at optimation this value isn't differentiable at zero. 


### Mean Absolute Percetang Error(MAPE)
--- same as above just divided by the original to make it a  percetage

--- this is more useful because the units are no longer defined, this is a percentage

### 2. Mean Squared Error (MSE)
--- the mean squared error and the rmse is related is that the rmse is the root of hte mean squared error. These are the go to for calculating the 'loss' between the predicted and the actual

--- MSE is better for optimization like for gradient descent thatn actually calculating fo loss
Definition: Average of squared differences between predicted and actual values.

Function: sklearn.metrics.mean_squared_error(y_true, y_pred)

Usage: Penalizes larger errors more strongly.



### 3. Root Mean Squared Error (RMSE)
Definition: Square root of MSE.

Usage: Same units as target variable, interpretable scale.



### 4. Mean Squared Logarithmic Error (MSLE)
Definition: Penalizes underestimation more than overestimation.

Function: sklearn.metrics.mean_squared_log_error(y_true, y_pred)

Usage: Useful when target values span several orders of magnitude.

### 5. R² Score (Coefficient of Determination)
Definition: Proportion of variance explained by the model.

Function: sklearn.metrics.r2_score(y_true, y_pred)

Usage: Measures goodness of fit; ranges from -∞ to 1.

-- DOESN"T say how good the model performs
-- THE NORMAL r2 wouldn't work in most models if implemented because the level of determination increases the more the features are added. The solution is to penalize

### 6. Explained Variance Score
Definition: Measures proportion of variance explained by predictions.

Function: sklearn.metrics.explained_variance_score(y_true, y_pred)

Usage: Similar to R² but less strict.

### 7. Median Absolute Error
Definition: Median of absolute errors.

Function: sklearn.metrics.median_absolute_error(y_true, y_pred)

Usage: Robust to outliers compared to MAE.
