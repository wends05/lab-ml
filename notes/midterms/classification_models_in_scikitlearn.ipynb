| Model                        | Key Hyperparameters                                   | Best Use Case                                |
|-------------------------------|-------------------------------------------------------|----------------------------------------------|
| **Logistic Regression**       | `penalty`, `C`, `solver`, `max_iter`                  | Linear decision boundary, interpretable       |
| **K-Nearest Neighbors (KNN)** | `n_neighbors`, `weights`, `algorithm`, `p`            | Instance-based learning, small datasets       |
| **Support Vector Classifier** | `kernel`, `C`, `gamma`, `degree`                      | High-dimensional spaces, non-linear boundaries|
| **Decision Tree**             | `criterion`, `max_depth`, `min_samples_split`, `min_samples_leaf` | Interpretability, non-linear splits          |
| **Random Forest**             | `n_estimators`, `max_features`, `max_depth`, `bootstrap` | Robust ensemble, reduces overfitting         |
| **Gradient Boosting**         | `n_estimators`, `learning_rate`, `max_depth`, `subsample` | High accuracy, boosting weak learners        |
| **AdaBoost**                  | `n_estimators`, `learning_rate`, `base_estimator`     | Ensemble boosting, handles weak classifiers  |
| **Naive Bayes (GaussianNB)**  | `var_smoothing`                                       | Text classification, probabilistic approach  |
| **BernoulliNB**               | `alpha`, `binarize`                                   | Binary/boolean features                      |
| **MultinomialNB**             | `alpha`, `fit_prior`                                  | Text classification with word counts         |
| **Linear Discriminant Analysis (LDA)** | `solver`, `shrinkage`, `n_components`        | Dimensionality reduction + classification    |
| **Quadratic Discriminant Analysis (QDA)** | `reg_param`                               | Non-linear class boundaries                  |
| **Perceptron**                | `penalty`, `alpha`, `max_iter`, `tol`                 | Simple linear classifier, online learning    |
| **SGDClassifier**             | `loss`, `penalty`, `alpha`, `max_iter`, `learning_rate` | Large-scale linear classification           |
| **Extra Trees Classifier**    | `n_estimators`, `max_features`, `criterion`, `max_depth` | Faster ensemble, randomized trees            |
