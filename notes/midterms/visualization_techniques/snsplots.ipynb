### Violin Plot
<small>A <b>violin plot</b> is a visualization that blends the features of a boxplot with a density plot. It shows the median and quartiles (like a boxplot), but also the full distribution shape of the data, so you can see skewness, multimodality, or spread more clearly.</small>

When to Use
- To compare distributions across categories (e.g., sales across outlet types).

- To detect skewness or multimodal patterns that boxplots alone might hide.

- To visualize spread and density more clearly than histograms when grouped by categories

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.violinplot(
    x="Item_Type",
    y="Item_Outlet_Sales",
    data=df,
    palette="Set2",
    inner="box"   # adds a mini boxplot inside each violin
)

plt.xticks(rotation=45)
plt.title("Distribution of Sales by Item Type")
plt.show()

# split by hue
sns.violinplot(x="Item_Fat_Content", y="Item_Outlet_Sales", hue="Outlet_Size", data=df, split=True)


# overlay with a swarmplot
sns.violinplot(x="Outlet_Type", y="Item_Outlet_Sales", data=df, inner=None, palette="pastel")
sns.swarmplot(x="Outlet_Type", y="Item_Outlet_Sales", data=df, color="k", alpha=0.5)

# focus only on sahpe of distribution
sns.violinplot(x="Item_Type", y="Item_Outlet_Sales", data=df, inner=None)

```

### Heatmap
<small>
A heatmap is a visualization that uses color intensity to represent values in a matrix. In data analysis, it’s often used to show correlation matrices or frequency tables, making patterns and relationships easier to spot at a glance.

Uses of Heatmaps
- Correlation analysis: Visualize how strongly variables are related.

- Frequency counts: Show how often combinations of values occur.

- Highlighting patterns: Spot clusters, trends, or anomalies in large datasets.

- Feature selection: Quickly identify variables that may be redundant or highly correlated.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Compute correlation matrix for numerical columns
corr = df[["Item_Weight", "Item_Visibility", "Item_MRP", "Item_Outlet_Sales"]].corr()

# Plot heatmap
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

### Pairplot
A pairplot is one of Seaborn’s most versatile tools because it lets you explore all pairwise relationships between variables in a dataset at once. It’s essentially a grid of plots: scatterplots off-diagonal and distribution plots (histograms or KDEs) on the diagonal.

<small>Exploratory Data Analysis (EDA)

Quickly visualize how variables relate to each other.

Spot correlations, clusters, or outliers.

Example: See how Item_MRP, Item_Visibility, and Item_Outlet_Sales interact.

Correlation Insight

Scatterplots show whether variables are positively/negatively correlated.

Diagonal plots show the distribution of each variable.

Category Comparison

Use hue to color points by a categorical variable (e.g., Item_Type).

Helps reveal differences in relationships across groups.

Regression Visualization

Set kind="reg" to add regression lines in scatterplots.

Useful for checking linear trends between variables.

Distribution Analysis

Change diagonal plots with diag_kind="kde" for smooth density curves.

Reveals skewness or multimodal distributions.

```python
# Basic pairplot
import seaborn as sns
import matplotlib.pyplot as plt

numerical_cols = ["Item_Weight", "Item_Visibility", "Item_MRP", "Item_Outlet_Sales"]

sns.pairplot(df, vars=numerical_cols)
plt.show()
```

```python
# pair plot with coloring
sns.pairplot(df, vars=numerical_cols, hue="Item_Type", palette="Set2")
plt.show()
```

```python
# pairplot with reg line
sns.pairplot(df, vars=numerical_cols, hue="Outlet_Size", kind="reg", diag_kind="kde")
plt.show()
```

pairplots are your “big picture” visualization tool — they let you see how every variable interacts with every other variable, while also showing each variable’s distribution.
