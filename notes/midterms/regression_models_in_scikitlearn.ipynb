| Model                  | Key Hyperparameters                  | Best Use Case                          |
|-------------------------|--------------------------------------|----------------------------------------|
| **Linear Regression**   | `fit_intercept`, `n_jobs`            | Simple linear relationships            |
| **Ridge**               | `alpha`, `solver`, `max_iter`        | Multicollinearity, shrinkage           |
| **Lasso**               | `alpha`, `max_iter`, `selection`     | Feature selection, sparsity            |
| **ElasticNet**          | `alpha`, `l1_ratio`, `max_iter`      | Balance between Ridge & Lasso          |
| **SVR**                 | `kernel`, `C`, `epsilon`, `gamma`    | Non-linear regression                  |
| **Decision Tree**       | `criterion`, `max_depth`, `min_samples_split`, `min_samples_leaf` | Interpretability, non-linear splits    |
| **Random Forest**       | `n_estimators`, `max_features`, `max_depth` | Robust ensemble regression             |
| **Gradient Boosting**   | `n_estimators`, `learning_rate`, `max_depth`, `subsample` | High accuracy, boosting                |
| **KNN**                 | `n_neighbors`, `weights`, `algorithm`, `p` | Local similarity-based regression      |
| **Bayesian Ridge**      | `alpha_1`, `alpha_2`, `lambda_1`, `lambda_2` | Probabilistic linear regression        |


### 1.Linear Regression (LinearRegression)
Hyperparameters:

fit_intercept: Whether to calculate the intercept.

normalize: Deprecated, use StandardScaler.

copy_X: Copy input data.

n_jobs: Number of jobs for computation.

### 2. Ridge Regression (Ridge)
Hyperparameters:

alpha: Regularization strength.

solver: Algorithm (auto, svd, cholesky, lsqr, etc.).

fit_intercept: Include intercept.

max_iter: Maximum iterations.

### 3. Lasso Regression (Lasso)
Hyperparameters:

alpha: Regularization strength.

max_iter: Maximum iterations.

tol: Tolerance for optimization.

selection: Coordinate descent strategy (cyclic, random).

### 4. ElasticNet (ElasticNet)
Hyperparameters:

alpha: Overall regularization.

l1_ratio: Balance between L1 and L2 penalty.

max_iter, tol, selection.

### 5. Support Vector Regression (SVR)
Hyperparameters:

kernel: (linear, poly, rbf, sigmoid).

C: Regularization parameter.

epsilon: Tube within which no penalty is given.

gamma: Kernel coefficient.

### 6. Decision Tree Regression (DecisionTreeRegressor)
Hyperparameters:

criterion: (squared_error, friedman_mse, absolute_error).

max_depth: Maximum depth of tree.

min_samples_split: Minimum samples to split.

min_samples_leaf: Minimum samples per leaf.

### 7. Random Forest Regression (RandomForestRegressor)
Hyperparameters:

n_estimators: Number of trees.

criterion: Function to measure split quality.

max_depth, min_samples_split, min_samples_leaf.

max_features: Number of features considered at each split.

### 8. Gradient Boosting Regression (GradientBoostingRegressor)
Hyperparameters:

n_estimators: Number of boosting stages.

learning_rate: Shrinks contribution of each tree.

max_depth: Depth of individual trees.

subsample: Fraction of samples used for fitting.

9. K-Nearest Neighbors Regression (KNeighborsRegressor)
Hyperparameters:

n_neighbors: Number of neighbors.

weights: (uniform, distance).

algorithm: (auto, ball_tree, kd_tree, brute).

p: Power parameter for Minkowski metric.

### 10. Bayesian Ridge Regression (BayesianRidge)
Hyperparameters:

alpha_1, alpha_2: Priors for alpha.

lambda_1, lambda_2: Priors for lambda.

fit_intercept, normalize
