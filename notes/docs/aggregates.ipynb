| Function | Description | Example |
|----------|-------------|---------|
| `sum`    | Total of values | `df['col'].aggregate('sum')` |
| `mean`   | Average value | `df.aggregate('mean')` |
| `min`    | Minimum value | `df.aggregate('min')` |
| `max`    | Maximum value | `df.aggregate('max')` |
| `count`  | Number of non-NaN values | `df.aggregate('count')` |
| `median` | Median value | `df.aggregate('median')` |
| `std`    | Standard deviation | `df.aggregate('std')` |
| `var`    | Variance | `df.aggregate('var')` |
| `prod`   | Product of values | `df.aggregate('prod')` |
| `first`  | First value | `df.aggregate('first')` |
| `last`   | Last value | `df.aggregate('last')` |
