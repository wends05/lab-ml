### K Nearest Neighbor

Intuition of KNN - You can further classify a cat into a domestic cat or a lion based of characteristics that are similar 

<small>Here it shows that in the data set, shows that domesitc cats have shorter ears and heights while lions are the opposite. KNN classifies the pet cat and the lion based on how close the data is to each other, when this input is given , tries to match the characcteristics of hte closess neighbors</small>

<small> K determines the amount of nearest neighbors to take into consideration, the unknown is classifie dby taking the majority of votes from k nearest neighbors </small>

How to choose K Values (Parameter Tuning)

<small>Classification is bassed on the number of majority votes on the neighbors </small>

<small> If you have an even number of samples, you should make k odd </small>

```python
# requires data to be scaled
import sklearn

from sklearn.neighbors import KNeighborsClassifier
import numpy as np

k = int(np.sqrt(len(y_test)))

knn = KNeighborsClassifier(n_neighbors=k, metric='euclidean')
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
```
