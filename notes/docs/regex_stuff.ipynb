# Essential Regex Patterns in Python

| **Pattern** | **Meaning** | **Example Match** |
|-------------|-------------|-------------------|
| `.`         | Any character except newline | `a.c` → matches `abc`, `axc` |
| `^`         | Start of string              | `^Hello` → matches `Hello world` |
| `$`         | End of string                | `world$` → matches `Hello world` |
| `\d`        | Digit (0–9)                  | `\d+` → matches `123` |
| `\D`        | Non-digit                    | `\D+` → matches `abc` |
| `\w`        | Word character (alphanumeric + `_`) | `\w+` → matches `hello_123` |
| `\W`        | Non-word character           | `\W+` → matches `@#!` |
| `\s`        | Whitespace (space, tab, newline) | `\s+` → matches `   ` |
| `\S`        | Non-whitespace               | `\S+` → matches `text` |
| `*`         | 0 or more repetitions        | `ab*c` → matches `ac`, `abc`, `abbc` |
| `+`         | 1 or more repetitions        | `ab+c` → matches `abc`, `abbc` |
| `?`         | 0 or 1 repetition            | `ab?c` → matches `ac`, `abc` |
| `{n}`       | Exactly n repetitions        | `a{3}` → matches `aaa` |
| `{n,m}`     | Between n and m repetitions  | `a{2,4}` → matches `aa`, `aaa`, `aaaa` |
| `[]`        | Character set                | `[aeiou]` → matches any vowel |
| `[^]`       | Negated set                  | `[^aeiou]` → matches consonants |
| `|`         | OR operator                  | `cat|dog` → matches `cat` or `dog` |
| `()`        | Grouping                     | `(ab)+` → matches `ab`, `abab` |
| `(?P<name>...)` | Named group              | `(?P<year>\d{4})` → captures year |
| `(?=...)`   | Positive lookahead           | `\d(?=px)` → matches digit before `px` |
| `(?!...)`   | Negative lookahead           | `\d(?!px)` → digit not followed by `px` |

---

##  Common Use Cases

- **Email validation**: `r'\b[\w.-]+@[\w.-]+\.\w{2,4}\b'`
- **Phone numbers**: `r'\+?\d{1,3}[-.\s]?\(?\d{2,4}\)?[-.\s]?\d{3,4}[-.\s]?\d{3,4}'`
- **Dates (YYYY-MM-DD)**: `r'\d{4}-\d{2}-\d{2}'`
- **Extract hashtags**: `r'#\w+'`
- **Split on commas outside quotes**: `r',(?=(?:[^"]*"[^"]*")*[^"]*$)'`

---

## Python Regex Workflow

```python
import re

pattern = r'\d{4}-\d{2}-\d{2}'  # date pattern
text = "Today's date is 2026-01-10."

match = re.search(pattern, text)
if match:
    print(match.group())  # Output: 2026-01-10


# 🔑 Word Boundary Regex Patterns

| **Pattern** | **Meaning** | **Example Match** |
|-------------|-------------|-------------------|
| `\b`        | Word boundary(EDGeS OF A WORD) (position between a word char `\w` and a non-word char `\W`) | `\bcat\b` → matches `"cat"` in `"black cat"`, but not `"concatenate"` |
| `\B`        | Non-word boundary(INSiDe A Word) (position NOT at a word boundary) | `\Bcat\B` → matches `"concatenate"`, but not `"cat"` alone |
