```python
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
#get datasets within seaborn
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset("penguins")
print(penguins.head())
```

```python
penguins['species'].value_counts()
```

```python
penguins['island'].value_counts()
```

ScatterPlot

```python
# data is of penguins, the x axis flipper length in mm, y axis bill length in mm, hue is species 
# you can set the hue for different colors for different species or islands
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species')
```

default styling themes

```python
# the way sof styling seaborn plots (using set_style) (backgound styles)
# white
# dark
# whitegridd
# darkgrid
# ticks

# the way to set label and size of the plot
#set_context can also be used to change the size of the plot allows to override default parameters
#poster - very large elements
#paper = ideal for academic paper
#notebook = ideal for jupyter notebooks
#talk - ideal for presentations
```

```python
sns.set_style("ticks")
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species')
sns.despine()  #removes the top and right spines
# can do sns.despine(left=True) to remove left spine also
sns. set_context("poster")
```

Pallete

```python
sns.set_style("ticks")
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species', palette='Set1')
sns.despine()  #removes the top and right spines
# can do sns.despine(left=True) to remove left spine also
```

using style to distinguish data points, size to distinguish size

```python
sns.scatterplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='island', style='sex', size='body_mass_g', sizes=(20, 200), alpha=0.7, palette='Set2')
```

Histogram

```python
sns.histplot(data=penguins, x='flipper_length_mm', kde=True, hue='sex', multiple='stack')
#kde makes a smooth curve over the histogram
#multiple can be 'stack' or 'dodge' or 'fill'
# stack - stacks the bars
# dodge - places the bars side by side
# fill - normalizes the bars to 100%
```

Regression Plot

```python
sns.regplot(data=penguins, x='flipper_length_mm', y='body_mass_g', scatter=True, fit_reg=True,)
```

Line Plot

```python
sns.set_style("whitegrid")
sns.lineplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species', errorbar=('ci', 95), style='sex')

#ci is confidence interval, default is 95, can set to None to remove the shaded area
```

Bar plot

```python
sns.barplot(data=penguins, x='species', y='body_mass_g', palette='Set3', hue='sex', estimator=np.median)
#estimator can be np.mean, np.median, len, sum etc.
#the estimator is used to calculate the height of the bars
```

Count Plot

```python
sns.countplot(data=penguins, x='species', hue='island', palette='Set2')
```

KDE Plot

<small>a smooth curve that shows a continuous distribution of data</small>

```python
sns.kdeplot(data=penguins, x='flipper_length_mm', hue='species', fill=True, common_norm=False, alpha=0.5, palette='Set1')
# fill=True fills the area under the curve
#common norm=False ensures that the area under each curve sums to 1
```

Heat Map

<small>shows relationship or pattern between two variables </small>

```python
columns = ['flipper_length_mm', 'bill_length_mm', 'bill_depth_mm', 'body_mass_g']

sns.heatmap(penguins[columns].corr(), annot=True, cmap='Blues')
#annot=True adds the correlation values on the heatmap (can remove to just show colors)
#cmap is the color map used for the heatmap#
#vmin and vmax can be used to set the range of values for the colormap
#cmap options: 'viridis', 'plasma', 'inferno', 'magma', 'cividis', 'coolwarm', 'YlGnBu', etc., 'Blues', 'Reds', 'Greens'
```
