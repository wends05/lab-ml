| Function / Method        | Works On                  | Purpose                                      | Returns                          | Example |
|--------------------------|---------------------------|----------------------------------------------|----------------------------------|---------|
| `filter(function, iterable)` | Any iterable              | Keeps only elements that meet a condition     | Iterator (convert with `list()`) | `list(filter(lambda x: x%2==0, [1,2,3,4])) → [2,4]` |
| `find()` (string method) | Strings                   | Finds index of a substring                   | Integer (`-1` if not found)      | `"hello".find("lo") → 3` |
| `replace()` (string method) | Strings                   | Replaces all occurrences of a substring      | New string                       | `"123-456".replace("-", "") → "123456"` |
| `re.sub(pattern, repl, string)` | Strings (regex)           | Replaces substrings matching a regex pattern | New string                       | `re.sub(r"\D", "", "123-456") → "123456"` |
| `map(function, iterable)` | Any iterable              | Applies a function to each element           | Iterator (convert with `list()`) | `list(map(lambda x: x**2, [1,2,3])) → [1,4,9]` |
| `contains (in)`          | Strings, lists, sets, dicts | Checks if an element/substring exists        | Boolean (`True`/`False`)         | `"cat" in "concatenate" → True`<br>`2 in [1,2,3] → True` |
