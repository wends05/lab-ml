Value Imputation

<small>Deleting a feature is never preferred as it is a feature that be correlated to an outcome. Usually, you use EDA or measures of central tendancy like mean, median mode or interpolation to supply null data , if null is missing or too much is dropped it becomes a problem in data generalization</small>

<small>SimpleInputter is a very convenient strategy for missing data, replaces all missing with statistcis (mean, median, mode (frequency)) from neighboring columns </small>

```python
from sklearn.impute import SimpleImputer
```

```python
# making a strategy for imputing missing values
imputer = SimpleImputer(strategy='mean')
df['age'] = imputer.fit_transform(df[['age']])
```

```python
df['age'].isnull().sum()
```

```python
def get_paramaters(df): 
    parameters = {}
    for col in df.columns[df.isnull().any()]:
       if df[col].dtype == 'float64' or df[col].dtype == 'int64' or df[col].dtype == 'float32' or df[col].dtype == 'int32':
           strategy = 'mean'
       else: 
           strategy = 'most_frequent'
           
       missing_values = df[col][df[col].isnull()].values[0]
       # store starategy and missing value representation for column
       parameters[col] = {'strategy': strategy, 'missing_values': missing_values}
    return parameters

get_paramaters(df)
```

```python
parameters = get_paramaters(df)
```

```python
for col, param in parameters.items():
    imputer = SimpleImputer(strategy=param['strategy'], missing_values=param['missing_values'])
    df[col] = imputer.fit_transform(df[[col]])[:, 0]
```

Data Encoding

```python
from sklearn.preprocessing import OneHotEncoder
# there are different types of encoders like OrdinalEncoder, LabelEncoder
# onehotencoder creates binary columns for each category in a categorical feature (good for x)
# labelencoder creates a single column with integer labels for each category (good for y)
```

```python
df[['male', 'female'] ] = OneHotEncoder().fit_transform(df[['sex']]).toarray()
# addidiontal bracket selects as 2d dataframe
df[['sex', 'male', 'female']]
# you can see that the original 'sex' column has been transformed into two new binary columns 'male' and 'female
```

```python
# convert sex column into the binary columns
# selectes the second columd with [:, 1]
df['sex'] = OneHotEncoder().fit_transform(df[['sex']]).toarray()[:, 1]
```
