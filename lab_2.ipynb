```python
# importer les bibliothèques nécessaires

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
# Les données sont chargées à partir d'un fichier CSV

# Randomizer la graine pour la reproductibilité
np.random.seed(0)

data_size = 20

data = {
    'Country': np.random.choice(['France', 'Spain', 'Germany'], data_size),
    'Age': np.random.randint(25, 50, data_size),
    'Salary': np.random.randint(30000, 90000, data_size),
    'Purchased': np.random.choice(['No', 'Yes'], data_size) # The Target Variable
}
df = pd.DataFrame(data)


# Afficher les premières lignes du DataFrame
display(df.head())
```

```python
# Créer les variables de caractéristiques (X) et la variable cible (y)
X = df.iloc[:, :-1]  # Features

y = df[
    'Purchased'
]

print(y.head())
print(X)
```

```python
# Machines ne comprends pas les textes comme "France", "Spain", "Germany", "Yes", "No"
# Nous devons les encoder en valeurs numériques
le = LabelEncoder()
y_encoded = le.fit_transform(y)

X_encoded = pd.get_dummies(X, columns=['Country'])

# Afficher drop_first=True pour éviter le piège des variables fictives
display(X_encoded.head())
```

```python
# Ne pas trainer sur tout le dataset pour éviter l'overfitting
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y_encoded, test_size=0.2, random_state=42)

print(f"Total Rows: {len(X_encoded)}")
print(f"Training Rows: {len(X_train)}")
print(f"Testing Rows: {len(X_test)}")
```

```python
# Feature Scaling
scaler = StandardScaler()

X_trained_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Convertir les tableaux numpy en DataFrames pour une meilleure lisibilité
X_train_final = pd.DataFrame(X_trained_scaled, columns=X_encoded.columns)

print("--- Scaled Training Features ---")
display(X_train_final.head())
# Notice: L'Age et le Salaire sont maintenant sur une échelle similaire
```

```python
# Vérifier les valeurs manquantes
# Creer un Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier()
clf = clf.fit(X_train_final, y_train)
```
