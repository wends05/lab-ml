Get the unique values in all columns

```python
for col in df.columns:
    print(f"Unique values in column '{col}': {df[col].unique()}\n")
```

Data contains certain values. cleanup needed.

```python
error_values = ['unknown', 'nan']

for col in df.columns:

    if df[col].dtype == 'object':
        df[col] = df[col].str.strip().str.lower()

    # replace the error values with NaN
    df[col] = df[col].replace(error_values, pd.NA)
    if df[col].dtype == 'float64' or df[col].dtype == 'int64':
        df[col] = pd.to_numeric(df[col], errors='coerce')
```
